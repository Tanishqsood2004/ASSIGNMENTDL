# 3. What is the primary function of tf.function in TensorFlow 2.0?


tf.function is used to convert a Python function into a TensorFlow graph, which improves performance.

```python
import tensorflow as tf

@tf.function
def add(a, b):
    return a + b
```

Without tf.function: runs eagerly (line by line).

With tf.function: runs as a compiled graph (faster and optimized).

# 4. What is the purpose of the Model class in TensorFlow 2.0?


The Model class (from tf.keras.Model) represents a complete model that can be trained, evaluated, and used for inference.

Purpose:
Encapsulates the layers, forward pass, training logic, and evaluation.

Helps in building custom models with full control over training.

Example:

```python
class MyModel(tf.keras.Model):
    def __init__(self):
        super(MyModel, self).__init__()
        self.dense1 = tf.keras.layers.Dense(64, activation='relu')
        self.out = tf.keras.layers.Dense(10)

    def call(self, inputs):
        x = self.dense1(inputs)
        return self.out(x)
```

Importance:
Every data point (input, output, weights, gradients) is a tensor.

Operations in TensorFlow are defined over tensors (like NumPy arrays but with GPU acceleration).

Understanding tensor shapes and ranks is key to designing and debugging models.

#6. How do you create a simple neural network in PyTorch?




```python
import torch.nn as nn

class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.fc1 = nn.Linear(4, 10)
        self.fc2 = nn.Linear(10, 1)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        return self.fc2(x)

model = Net()
```
